### Creating the Exercise Datasets

This notebook creates three DataFrames: employee details, seniority levels, and project assignments. The exercise includes five employee records and 14 project records. The DataFrames are saved as Employee.csv, Seniority.csv, and Project.csv, then read back for further processing.


```python
# Creating Employee dataframe
employee = pd.DataFrame({
    'ID':['A001','A002','A003','A004','A005'],
    'Name':['John Alter','Alice Luxumberg','Tom Sabestine','Nina Adgra','Amy Johny'],
    'Gender':['M','F','M','F','F'],
    'City':['Paris','London','Berlin','Newyork','Madrid'],
    'Age':[25,27,29,31,30]
})

print(employee)
```

```python
# Creating Seniority dataframe
seniority = pd.DataFrame({
    'ID':['A001','A002','A003','A004','A005'],
    'Designation Level':[2,2,3,2,3]
})

print(seniority)
```

```python
# Creating Project dataframe
project = pd.DataFrame({
    'ID':['A001','A002','A003','A004','A005',
          'A002','A005','A003','A001','A003',
          'A001','A004','A004','A005'],

    'Project':['Project1','Project2','Project3','Project4','Project5',
               'Project6','Project7','Project8','Project9','Project10',
               'Project11','Project12','Project13','Project14'],

    'Cost':[1002000,2000000,4500000,5500000,np.nan,
            680000,400000,350000,np.nan,300000,
            2000000,1000000,3000000,200000],

    'Status':['Finished','Ongoing','Finished','Ongoing','Finished',
              'Failed','Finished','Failed','Ongoing','Finished',
              'Failed','Ongoing','Finished','Finished']
})

print(project)
```

```python
# Saving dataframes as CSV files
employee.to_csv('Employee.csv',index=False)
seniority.to_csv('Seniority.csv',index=False)
project.to_csv('Project.csv',index=False)
```

### Handling Missing Project Costs

Missing project costs are filled using a running average. The notebook maintains a running sum and count of previously available costs. When a cost is missing, it uses the average of the previously encountered non-missing costs. For example, the missing cost in the fifth project record is filled using the average of the first four available costs.


```python
# Reading project file
project = pd.read_csv('Project.csv')

# Creating running sum and count
running_sum = 0
count = 0

for i in range(len(project)):

    if pd.notnull(project.loc[i,'Cost']):
        running_sum += project.loc[i,'Cost']
        count += 1

    else:
        project.loc[i,'Cost'] = running_sum/count

print(project)
```

### Data Transformation and Merging

The employee Name column is split into First Name and Last Name, and the original Name column is removed. The employee DataFrame is then merged with the seniority and project DataFrames using the employee ID. This combines employee details, designation levels, and project information for analysis.


```python
# Reading employee file
employee = pd.read_csv('Employee.csv')

# Splitting Name column
employee[['First Name','Last Name']] = employee['Name'].str.split(' ',expand=True)

# Removing old Name column
employee.drop('Name',axis=1,inplace=True)

print(employee)
```

```python
# Merge employee and seniority
final = pd.merge(employee,seniority,on='ID')

# Merge with project dataframe
final = pd.merge(final,project,on='ID')

print(final)
```

### Bonus Calculation

The notebook uses NumPy's `where()` function to calculate a bonus for each project record. Projects marked Finished receive a bonus equal to 5% of the project cost. Projects with any other status receive a bonus of zero. The calculation is performed at the project-record level.


```python
# Creating Bonus column
final['Bonus'] = np.where(
    final['Status']=='Finished',
    final['Cost']*0.05,
    0
)

print(final)
```

```python
# Reduce designation by 1 for failed projects
final.loc[
    final['Status']=='Failed',
    'Designation Level'
] += 1

# Delete employees whose designation is above 4
final = final[final['Designation Level']<=4]

print(final)
```

```python
# Adding title
final['First Name'] = np.where(
    final['Gender']=='M',
    'Mr. '+final['First Name'],
    'Mrs. '+final['First Name']
)

# Dropping Gender
final.drop('Gender',axis=1,inplace=True)

print(final)
```

```python
# Increase designation for employees above age 29
final.loc[
    final['Age']>29,
    'Designation Level'
] -= 1

print(final)
```

### Grouping and Analyzing Project Costs

The merged data can be grouped by employee to summarize project costs. The saved grouped output assigns employee A004 a total project cost of 9,500,000. This is a total from the exercise data; the currency or cost unit is not specified in the screenshots.

```python
# Group by ID and First Name
totalprojcost = final.groupby(
    ['ID','First Name']
)['Cost'].sum().reset_index()

# Rename column
totalprojcost.rename(
    columns={'Cost':'Total Cost'},
    inplace=True
)

print(totalprojcost)
```
